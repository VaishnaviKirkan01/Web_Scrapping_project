```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import pandas as pd
df=pd.read_csv("C:\\Users\\vaish\\OneDrive\\Documents\\Web_Scrapping_project\\Cleaned_Laptops.csv")
```

```python
sns.histplot(
    x=df['price'],
    bins=20,
    color='pink'
)
plt.title("Laptop Price Histogram")
plt.xlabel("Price")
plt.ylabel("Number of laptops")
plt.legend(['Price'])
plt.show()
```

```python
sns.countplot(
    x=df["brand"],
    color="green",
    
)
plt.title("The count of laptop by Brand")
plt.xlabel("Brand")
plt.ylabel("Count")
plt.legend(["count"])
plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(
    df["brand"],
    df['price'],
    color="orange",
    
)

plt.title("Price By Brand")
plt.xlabel('Brand')
plt.ylabel("Price")
plt.legend(["price"])
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxenplot(
    x=df['brand'],
    y=df['ram'],
    palette="muted"
)

plt.title("Brand vs RAM")
plt.title("Brands")
plt.ylabel("RAM")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(
    x=df['ssd_storage'],
    palette="deep"

)

plt.title("SSD Distribution")
plt.legend(['count'])
plt.xlabel("Storage")
plt.ylabel('Number of laptops')
plt.show()
```

```python
plt.scatter(
    df['ram'],
    df['price'],
    color="purple"
)

plt.title("Price vs RAM")
plt.xlabel("RAM")
plt.ylabel("Price")
plt.show()
```

```python
sns.barplot(
    data=df,
    x=df["ssd_storage"],
    y=df['price'],
    palette='pastel',
    hue='brand'
)

plt.title("Price vs SSD Storage")
plt.xlabel("SSD_Storage")
plt.ylabel("Price")
plt.show()
```

```python
sns.histplot(
    data=df,
    x=df['rating'],
    bins=10,
    palette="deep"
)

plt.title("Laptop Rating Distribution")
plt.xlabel("rating")
plt.ylabel("No.of Laptops")
plt.show()
```

```python
sns.barplot(
    data=df,
    x=df['brand'],
    y=df['rating'],
    palette="pastel",

)

plt.title("Rating Vs Brand")
plt.xlabel("Brand")
plt.ylabel("rating")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.scatter(
    df['brand'],
    df["processor"],
    color="purple",

)
plt.title("Brand VS Processor")
plt.xlabel("Brand")
plt.ylabel("processor")
plt.xticks(rotation=45)
plt.show()
```
