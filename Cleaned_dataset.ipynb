```python
# converting the numerical columns into the numeric

df['rating']=pd.to_numeric(df['rating'],errors="coerce")
```

```python
#  cleane the price column
df['price']=df['price'].str.replace({",":'','₹':""})
df['price']=pd.to_numeric(df['price'],errors='coerce')
```

```python
# checking their data types
print('price data type= ',df['price'].dtypes)
print('rating data type= ',df['rating'].dtypes)
```

```python
df
```

```python
# clean ram column

df['ram']=df['ram'].str.replace("GB",'')
df['ram']=pd.to_numeric(df['ram'],errors="coerce")

df['ram'].unique()


```

```python
# cleaning the storage column
def sto(i):
    if isinstance(i,str):
        if "TB" in i:
            i=i.replace("TB",'')
            return int(i)*1024
        elif "GB" in i:
            i=i.replace("GB",'')
            return int(i)
        else:
            return i

df['ssd_storage']=df['ssd_storage'].apply(sto)


# df['ssd_storage']=pd.to_numeric(df['ssd_storage'],errors="coerce")


```

```python
# fill the missing values in ram and ssd storage
ram_median=df['ram'].median()
df['ram']=df['ram'].fillna(ram_median)

ssd_median=df["ssd_storage"].median()
df['ssd_storage']=df['ssd_storage'].fillna(ssd_median)
```

```python
# convert the ram amd storage column into int

df[["ram",'ssd_storage']]=df[["ram",'ssd_storage']].astype("int")
```

```python
df['title'].dropna(inplace=True)
```

```python
# converting the brand names in all first word capital

df['brand'].unique()
df['brand']=df['brand'].str.strip().str.title()
```

```python
# strip the extra spaces from column names

df.columns=df.columns.str.strip()
df.columns
```

```python
# fill the nan values int the color column

df['color']=df['color'].fillna("Unknown")
```

```python
# replace the unvalid spelling in color column

df['color']=df['color'].str.replace({
    "Grey":"Gray"
})
```

```python
# processsor column cleaning

# fill the nan values

df['processor']=df['processor'].fillna("Unknown")

# now all the column values in title foray

df["processor"]=df['processor'].str.strip().str.title()
```
